```python
Q1 = price["Avg_Modal_Price"].quantile(0.25)
Q3 = price["Avg_Modal_Price"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print(lower)
print(upper)
```

```python
outliers = price[
    (price["Avg_Modal_Price"] < lower) |
    (price["Avg_Modal_Price"] > upper)
]

print("Outliers:", len(outliers))
print("Percentage:", len(outliers) / len(price) * 100)
```

Target Variable

State-wise Insights

Goa, Kerala, and Mizoram have the highest average crop prices.
State is expected to be an important predictive feature.

Seasonality

Prices exhibit seasonal variation, with higher averages during August–November.
Month should be encoded using cyclical features (Month_sin, Month_cos) instead of using the raw month number.

Weather Analysis

Weather variables show weak linear correlations with price.
Non-linear models (e.g., XGBoost, LightGBM, CatBoost) are expected to capture more complex relationships.

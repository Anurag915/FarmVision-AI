We found that rainfall values as NaN due to mismatch in the timestamp. To avoid this problem, we first normalise the data individually and then merge them using pandas.merge rather than xarray.merge

```python
print(ds_avgua.valid_time.values[:5])
```

```python
print(ds_avgad.valid_time.values[:5])
```

Convert Both to DataFrames


```python
df_avgua = ds_avgua.to_dataframe().reset_index()
df_avgad = ds_avgad.to_dataframe().reset_index()
```

Normalize the Date

```python
df_avgua["valid_time"] = pd.to_datetime(df_avgua["valid_time"]).dt.normalize()
df_avgad["valid_time"] = pd.to_datetime(df_avgad["valid_time"]).dt.normalize()
```

Merge Using Pandas

```python
weather_df = pd.merge(
    df_avgua,
    df_avgad[["valid_time", "latitude", "longitude", "tp"]],
    on=["valid_time", "latitude", "longitude"],
    how="inner"
)
```

```python
weather_df.head()
```

```python
weather_df["Temperature_C"] = weather_df["t2m"] - 273.15
weather_df["DewPoint_C"] = weather_df["d2m"] - 273.15
weather_df["Rainfall_mm"] = weather_df["tp"] * 1000
```

```python
weather_df["Wind_Speed"] = np.sqrt(
    weather_df["u10"]**2 +
    weather_df["v10"]**2
)
```

```python
weather_df["Surface_Pressure_hPa"] = weather_df["sp"] / 100
```

```python
weather_df.head()
```

```python
weather_df.columns.tolist()
```

```python
weather_df.drop(
    columns=[
        "t2m",
        "d2m",
        "tp",
        "sp",
        "u10",
        "v10",
        "number",
        "expver"
    ],
    inplace=True
)
```

```python
weather_df.rename(columns={
    "valid_time": "Date",
    "latitude": "Latitude",
    "longitude": "Longitude"
}, inplace=True)
```

```python
weather_df["Year"] = weather_df["Date"].dt.year
weather_df["Month"] = weather_df["Date"].dt.month
```

```python
weather_df = weather_df[
    [
        "Date",
        "Year",
        "Month",
        "Latitude",
        "Longitude",
        "Temperature_C",
        "DewPoint_C",
        "Rainfall_mm",
        "Wind_Speed",
        "Surface_Pressure_hPa",
    ]
]
```

```python
weather_df["Month_Name"] = weather_df["Date"].dt.month_name()
```

```python
weather_df = weather_df[
    [
        "Date",
        "Year",
        "Month",
        "Month_Name",
        "Latitude",
        "Longitude",
        "Temperature_C",
        "DewPoint_C",
        "Rainfall_mm",
        "Wind_Speed",
        "Surface_Pressure_hPa",
    ]
]
```

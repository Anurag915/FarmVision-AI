### Drop Duplicates

```python
price_df = price_df.drop_duplicates()
```

### Stats of State, Year, Month, Standard_Crop

```python
price_df.groupby(
    ["State","Year","Month","Standard_Crop"]
).size().describe()
```

### Aggregation of Modal Price by grouping states, year, months, standard crop

```python
monthly_price = (
    price_df
    .groupby(
        ["State", "Year", "Month", "Standard_Crop"]
    )
    .agg(
        Avg_Modal_Price=("Modal_Price", "mean"),
        Median_Modal_Price=("Modal_Price", "median"),
        Price_STD=("Modal_Price", "std"),
        Min_Modal_Price=("Modal_Price", "min"),
        Max_Modal_Price=("Modal_Price", "max"),
        Market_Records=("Market", "count"),
        Unique_Markets=("Market", "nunique")
    )
    .reset_index()
)
```

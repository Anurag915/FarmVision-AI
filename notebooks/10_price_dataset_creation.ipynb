```python
import pandas as pd
import numpy as np
```

```python
price = pd.read_parquet(
    "../datasets/processed/monthly_crop_prices.parquet"
)

weather = pd.read_parquet(
    "../datasets/processed/state_monthly_weather.parquet"
)
```

```python
weather = weather.rename(
    columns={
        "State_Name": "State"
    }
)
```

```python
weather_mapping = {
    "Andaman & Nicobar": "Andaman And Nicobar Islands",
    "Chhattishgarh": "Chhattisgarh",
    "Tamilnadu": "Tamil Nadu",
    "Telengana": "Telangana",
    "Jammu and Kashmir": "Jammu And Kashmir"
}

weather["State"] = weather["State"].replace(weather_mapping)
```

```python
price_mapping = {
    "Andaman and Nicobar": "Andaman And Nicobar Islands",
    "Chattisgarh": "Chhattisgarh",
    "Jammu and Kashmir": "Jammu And Kashmir",
    "NCT of Delhi": "Delhi",
    "Pondicherry": "Puducherry",
    "Uttrakhand": "Uttarakhand"
}

price["State"] = price["State"].replace(price_mapping)
```

```python
print(sorted(set(price["State"]) - set(weather["State"])))

print(sorted(set(weather["State"]) - set(price["State"])))
```

```python
price_weather = price.merge(
    weather,
    on=["State", "Year", "Month"],
    how="left"
)

price_weather.isnull().sum()
```

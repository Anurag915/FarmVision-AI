### Cleaning of Dataset with missing rows


```python
missing_percentage = (df["Production"].isnull().sum() / len(df)) * 100
print(f"Missing Production: {missing_percentage:.2f}%")
```

```python
df = df.dropna(subset=["Production"])
```

```python
df.shape
```

```python
df.columns
```

```python
df.drop(columns=["index"], inplace=True)
```

```python
duplicates = df.duplicated().sum()
print("Duplicate rows:", duplicates)
```

```python
df["State_Name"].unique()
```

## Remove Trailing spaces from the categorical columns 

```python
text_columns = [
    "State_Name",
    "District_Name",
    "Season",
    "Crop"
]

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
df["State_Name"].unique()
```

```python
# Capitalise the first letter

text_columns = [
    "State_Name",
    "District_Name",
    "Season",
    "Crop"
]

for col in text_columns:
    df[col] = df[col].str.title()
```

```python
for col in text_columns:
    print("=" * 40)
    print(col)
    print(df[col].nunique())
    print(sorted(df[col].unique())[:20])
```

```python
def clean_price_file(df):
    """
    Clean a single yearly crop price dataset.
    """

    # -----------------------------
    # Convert Arrival Date
    # -----------------------------
    df["Arrival_Date"] = pd.to_datetime(df["Arrival_Date"])

    # -----------------------------
    # Remove rows with missing Max Price
    # -----------------------------
    df = df.dropna(subset=["Max_Price"])

    # -----------------------------
    # Remove duplicate rows
    # -----------------------------
    # Only remove duplicates if present
    
    # dup_count = df.duplicated().sum()
    # if dup_count > 0:
    #     print(f"Removing {dup_count} duplicate rows")
    #     df = df.drop_duplicates()


    # -----------------------------
    # Create time features
    # -----------------------------
    df["Year"] = df["Arrival_Date"].dt.year
    df["Month"] = df["Arrival_Date"].dt.month
    df["Day"] = df["Arrival_Date"].dt.day
    df["Quarter"] = df["Arrival_Date"].dt.quarter

    # -----------------------------
    # Strip unwanted spaces
    # -----------------------------
    text_columns = [
        "State",
        "District",
        "Market",
        "Commodity",
        "Variety",
        "Grade"
    ]

    for col in text_columns:
        df[col] = df[col].str.strip()

    return df
```

```python
clean_df = clean_price_file(df)

clean_df.info()
```

```python
import gc
import os

output_folder = "../datasets/cleaned/crop_price"

os.makedirs(output_folder, exist_ok=True)

for file in files:

    print(f"Cleaning {os.path.basename(file)}")

    df = pd.read_parquet(file)

    df = clean_price_file(df)

    output_file = os.path.join(
        output_folder,
        os.path.basename(file)
    )

    df.to_parquet(output_file, index=False)

    del df
    gc.collect()

print("All files cleaned successfully!")
```
